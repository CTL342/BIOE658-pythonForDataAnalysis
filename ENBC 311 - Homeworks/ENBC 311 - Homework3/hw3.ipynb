````python
import random
X = random.randrange(1, 11)

def ex2_func(num):
    if type(num) is int:
        return(2*num**2 - 5*num + 3)
    else:
        return None
    
def ex2_derivative(num):
    if type(num) is int:
        return (ex2_func(num + 1) - ex2_func(num)) / ((num + 1) - num)
    else:
        return None

print("Randomly generated number: ", X)
print("Y value at randomly generated number: ", ex2_func(X))
print("Forward difference discrete derivative: ", ex2_derivative(X))
````

# Ex 3
Suppose you have a list of words.  

Write a function that counts how many words start with a **vowel** (a, e, i, o, u) and how many start with a **consonant**.  

Test it with at least two different lists of words.


````python
words1 = ["Hi", "my", "name", "is", "Christopher", "Lee", "and", "I", "like", "to", "program", "in", "Python"]
words2 = ["I", "belive", "that", "grades", "matter", "a", "lot", "when", "you", "are", "applying", "to", "jobs", "if", "you", "know", "what", "I", "am", "trying", "to", "say"]

def words_start_vowel(words):
    vowels = ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']
    count = 0
    
    if type(words) is list:
        # Compares whether or not the first character of the word pulled from the list of words is in the 
        # temporary list of vowels created. 
        for word in words:
            if word[0] in vowels:
                 count += 1
        return count
    else:
        return None

print("Number of words that start with a vowel in first list of words: ", words_start_vowel(words1))
print("Number of words that start with a vowel in second list of words: ", words_start_vowel(words2))
````

# Ex 4
Write a program that computes factorials.

1. Write a function `fact_iter(n)` that computes factorial **iteratively**.  
2. Write a function `fact_rec(n)` that computes factorial **recursively**.  
3. Ask the user for a number and print both results.  

*Note*: Return `None` if `n < 0`.


````python
def fact_iter(n):
    sum = 0
    for i in range(int(n) + 1):
        # Edge case to make sure that the sum is not multiplied by zero
        if i == 0:
            sum += 1
        else:
            sum = sum * i
    return sum

def fact_rec(n):
    # Base case to ensure that the function eventually stops
    if(int(n) == 2):
        return 2
    # The recursive portion that multiplies the current number by the itself minues one
    # which is the definition of a factorial
    else:
        return int(n) * fact_rec(int(n) - 1)


num1 = input("Enter a number: ")
num2 = input("Enter another number: ")

print(f"Factorial iteratively for the number {num1}:", fact_iter(num1))
print(f"Factorial recursively for the number {num2}:", fact_rec(num2))
````

# Ex 5: Bubble Sort Algorithm

We want to implement the **Bubble Sort algorithm** to sort a list of numbers in ascending order.

### Bubble Sort (pseudocode)

```text
Algorithm BubbleSort(A):
    n ← length(A)
    repeat
        swapped ← false
        for i from 0 to n-2 do
            if A[i] > A[i+1] then
                swap A[i] and A[i+1]
                swapped ← true
        end for
    until swapped = false
```


### Tasks
1. Implement a function `bubble_sort(numbers)` that takes a list and returns a new sorted list.  
2. Also modify your function to optionally sort in **descending order** if the user provides a keyword argument `reverse=True`.  
3. Test your function with:  
   - `[5, 3, 8, 2, 1]`  
   - `[10, -2, 0, 7, 4]`  


````python
def bubble_sort(temp_arr, reverse):
    # Length of list is obtained to loop a certain amount of iterations
    l = len(temp_arr)
    # First for loop is to allow the list of numbers to undergo the sorting
    for i in range(l):
        # Second for loop swaps the current number with the next number if it is either greater or smaller
        # based on the value of the reverse parameter
        for j in range(l):
            # This is a edgecase to make sure that the code does not try to grab a value that does not exist
            # within a current index
            if j != l - 1:
                # Determines whether the current number is swapped with the next number whether it is smaller
                # or greater
                if reverse is False:
                    # Condition and variables are used to swap values within the list if the condition is
                    # satisfied
                    if temp_arr[j] > temp_arr[j + 1]:
                        temp = temp_arr[j]
                        temp_arr[j] = temp_arr[j + 1]
                        temp_arr[j + 1] = temp
                else:
                    if temp_arr[j] < temp_arr[j + 1]:
                        temp = temp_arr[j]
                        temp_arr[j] = temp_arr[j + 1]
                        temp_arr[j + 1] = temp
    return temp_arr

nums1 = [5, 3, 8, 2, 1]
nums2 = [10, -2, 0, 7, 4]

print("Ascended bubble sort for nums1 list of numbers: ", bubble_sort(nums1, False))
print("Ascended bubble sort for nums2 list of numbers: ", bubble_sort(nums2, False))
print("Descended bubble sort for nums1 list of numbers: ", bubble_sort(nums1, True))
print("Descended bubble sort for nums2 list of numbers: ", bubble_sort(nums2, True))
````
