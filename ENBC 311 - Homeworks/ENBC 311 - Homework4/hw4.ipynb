````python
# question 1
teams = {}
results = {}
file_path = '/Users/christopherlee/Desktop/Anaconda Workspace/Homework3 - ENBC 311/'

print("Last Week League Results")
print("------------------")
print("Team Name | Points")
print("------------------")
with open(file_path + 'league_last_week.txt', 'r') as file:
    next(file)
    
    for line_number, line in enumerate(file, 1):
        parts = line.strip().split()

        if parts:
            points = parts[-1]
            team_name_parts = parts[:-1]

            team_name = " ".join(team_name_parts)
            teams[team_name] = int(points)
            print(team_name + ": " + points)

print()
print("This Week League Results")
print("------------------")
print("Team Name | Outcome")
print("------------------")
with open(file_path + 'results_this_week.txt', 'r') as file:
    next(file)
    
    for line_number, line in enumerate(file, 1):
        parts = line.strip().split()

        if parts:
            outcome = parts[-1]
            team_name_parts = parts[:-1]

            team_name = " ".join(team_name_parts)
            results[team_name] = outcome
            print(team_name + ": " + outcome)
````

````python
# question 2
for team in results:
    if results[team] == "win":
        teams[team] = int(teams[team]) + 3
    elif results[team] == "tie":
        teams[team] = int(teams[team]) + 1

print()
print("Next Week League Results")
print("------------------")
print("Team Name | Points")
print("------------------")
with open(file_path + 'leauge_next_week.txt', 'w') as file:
    file.write("Teams	Pts\n")

    for team in teams:
        file.write(f"{team}: {teams[team]}\n")
        print(f"{team}: {teams[team]}")
        
````

````python
# Question 2
def compare_numbers(x, y):
    if x > y:
        return "x is greater"
    elif x < y:
        return "y is greater"
    else:
        return "x and y are equal"
````

````python
# Question 3
def find_max(nums):
    if len(nums) == 0:
        return None
        
    max_nums = list()
    max_nums_index = list()
    max_num = float('-inf')
    for index, num in enumerate(nums):
        if num > max_num:
            max_nums = list()
            max_nums_index = list()
            
            max_num = num
            max_nums.append(num)
            max_nums_index.append(index)
            
        elif num == max_num:
            max_nums.append(num)
            max_nums_index.append(index)
    return max_nums, max_nums_index
````
