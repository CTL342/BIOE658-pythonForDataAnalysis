## Problem 1: Sales Data Analysis

You are given a dataset called "sales_data.csv" that contains information about the sales of a company in the last quarter. The data contains the following columns:

- `Date`: The date of the sale in the format "MM/DD/YYYY"
- `Product`: The name of the product sold
- `Units Sold`: The number of units sold
- `Price per Unit`: The price of each unit in USD
- `Total Sales`: The total amount of sales in USD

You need to perform the following tasks using Pandas:

1. Load the dataset into a Pandas DataFrame using the `.read_csv()` function.
2. Convert the "Date" column to a Pandas datetime format using the `.to_datetime()` function.
3. Replace the 'Na' strings with np.nan so that Pandas will know to treat these as NAN values
4. Convert the columns containing numerics to floats (you may wish to use the Pandas .to_numeric() function)
5. Drop the "Total Sales" column from the DataFrame using the `.drop()` function.
6. Replace any NaN values in the "Price per Unit" column with the mean of the column using the `.fillna()` function.
7. Create a new column called "Revenue" that calculates the total revenue of each sale using the "Units Sold" and "Price per Unit" columns. 
8. Filter the DataFrame to only include rows where the "Product" column contains the word "Shirt". Use the `.str.contains()` function to accomplish this.
9. Create a bar chart showing the total number of units sold for each product. Use the `.plot()` function to create the chart.

Your final Python script should print out the following information:

- The number of rows and columns in the DataFrame
- The mean, standard deviation, minimum, and maximum of the "Units Sold" column
- The total revenue generated by the company in the last quarter
- A bar chart showing the total number of units sold for each product.


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Part 1
df = pd.read_csv('sales_data.csv')
print(df)
# Part 2
df.loc[:, 'Date'] = pd.to_datetime(df['Date'])
# Part 3
df.loc[df['Units Sold'] == "Na", 'Units Sold'] = np.nan
# Part 4
for col in ['Units Sold', 'Price per Unit', 'Total Sales']:
    df.loc[:, col] = pd.to_numeric(df[col], errors='coerce')
# Part 5
df = df.drop('Total Sales', axis='columns')
# Part 6
df['Units Sold'] = df['Units Sold'].fillna(df['Units Sold'].mean())
df['Price per Unit'] = df['Price per Unit'].fillna(df['Price per Unit'].mean())
# Part 7
df['Revenue'] = df['Price per Unit'] * df['Units Sold']
# Part 8
df = df[df['Product'].str.contains("Shirt")]
# Part 9
plt.bar(df['Product'], df['Units Sold'])
plt.xlabel('')
plt.ylabel('Units Sold')
plt.title('Units Sold for Each Product')
plt.show()

print("\nCompany Overview:")
print("Shape of Data (Rows, Columns):", df.shape)
print("Information on Units Sold:\n", df['Units Sold'].describe())
print("Total Revenue Last Quarter", df['Revenue'].sum())
```
