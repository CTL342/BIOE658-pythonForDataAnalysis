# **Problem 1: Predicting Blood Glucose Using Wearable Sensor Data**

Wearable health devices (such as Fitbit, Apple Watch, and Biostrap) measure several physiological signals, including:

- Heart rate (bpm)  
- Skin temperature (°C)  
- Galvanic Skin Response (GSR, a stress indicator)  
- Step count during the measurement period  

These types of measurements are commonly used in biomedical engineering research to estimate physiological variables and support real-time health monitoring.  

In this assignment, you will use the dataset `wearable_glucose_dataset.csv` to build a Keras-based regression model that predicts blood glucose (mg/dL) from wearable sensor measurements including heart rate, skin temperature, galvanic skin response (GSR), and step count. You will load and inspect the dataset, select features and the target, split the data into training and testing sets, build and train a neural network model, evaluate performance using MAE, RMSE, and R², and plot true vs. predicted glucose values to assess model reliability.
