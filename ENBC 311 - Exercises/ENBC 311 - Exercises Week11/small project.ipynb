# Project: Analyzing Car Sales Data
## Problem statement:
You are given a dataset containing information about car sales. Your task is to perform some basic data analysis and answer some questions about the data.

## Data:

The dataset are in 'car_sales.csv' file that contains the following columns:

* `Make`: the make of the car (e.g. Toyota, Honda, etc.)
* `Model`: the model of the car (e.g. Camry, Civic, etc.)
* `Year`: the year the car was made
* `Price`: the price of the car in dollars
* `Mileage`: the mileage of the car in miles
* `Color`: the color of the car
* `Transmission`: the type of transmission (e.g. manual, automatic, etc.)

## Questions:
Answer the following questions using the dataset:

1. What is the average price of cars in the dataset?
2. What is the most common color of cars in the dataset?*
3. What is the average mileage of cars with manual transmission?
4. What is the average price of Toyota cars in the dataset?
4. What is the most expensive car in the dataset? What is its make and model?
