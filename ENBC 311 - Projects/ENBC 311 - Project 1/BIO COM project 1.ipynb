##  Why Is This Important?

In biomedical engineering and biocomputational engineering, we often take measurements that depend on **energy** of the x-ray photon (like lights but are invisible) for example:

- X-ray detectors measuring photons at different energies  
- Spectral CT scanners  
- Energy-dispersive X-ray diffraction (EDXRD) systems  

Different materials (like bone, soft tissue, or contrast agents) produce **different energy-dependent signatures**.  
These signatures look like curves showing how much intensity we get at each energy.

For example:

- Material 1 has a profile: intensity vs. energy  
- Material 2 has another profile  
- A **mixture** of the two materials produces a measured profile that is a combination of both  

Our goal is to figure out **how much of each material is in the mixture**.

## The Basic Idea

Suppose we have:

- Material 1 (soft tissue) with profile  
  $$P_1(E)$$  
- Material 2 (Adipose) with profile  
  $$P_2(E)$$  

These functions tell us the **intensity** measured at energy \(E\).

Now imagine we measure a sample that contains **both materials**.  
The detector gives us a mixture profile:

- Mixture profile  
  $$P_m(E)$$

If the mixture contains:

- $a_1$ fraction of Material 1  
- $a_2$ fraction of Material 2

Those are the volumetric fraction of each material.

then a simple linear model describes the mixture:

$$
P_m(E) = a_1 P_1(E) + a_2 P_2(E)
$$

This makes sense:  
the total signal we detect is the sum of the signals from each material.

---

## But There Is a Problem…

Each material may have its profile measured at **different energy values**.

Example:

- Material 1 measured at energies  
  $$[20, 25, 30, 35, 40]$$
- Material 2 measured at  
  $$[22, 28, 33, 39]$$
- Mixture measured at  
  $$[20, 25, 30, 40]$$

To combine the profiles or solve the equation  
$$P_m = a_1 P_1 + a_2 P_2,$$  
all curves must be defined at the **same energy points**.

This is where **interpolation** becomes essential.

---

## Using Interpolation

We will use interpolation to:

1. Take each material’s profile  
2. Estimate its intensity values at the **common energy grid** used for the mixture  
3. Build a system of equations  
4. Solve for the unknown fractions $a_1$ and $a_1$

---

## After Interpolation: Solving for the Material Fractions

Once everything is on the same energy axis, we can stack the interpolated curves to form a matrix:

$$
\begin{bmatrix}
P_{m}(E_1) \\
P_{m}(E_2) \\
\vdots \\
P_{m}(E_n)
\end{bmatrix}
=
\begin{bmatrix}
P_1(E_1) & P_2(E_1) \\
P_1(E_2) & P_2(E_2) \\
\vdots & \vdots \\
P_1(E_n) & P_2(E_n)
\end{bmatrix}
\begin{bmatrix}
a_1 \\
a_2
\end{bmatrix}
$$

We solve this using **least squares** estimation.
